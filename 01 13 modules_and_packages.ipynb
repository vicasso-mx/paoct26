# ¿Qué problema resuelve?

Cuando un proyecto crece, colocar todo el código en un único archivo provoca:

- Archivos difíciles de mantener.
- Código duplicado.
- Baja reutilización.
- Mayor probabilidad de errores.

Los módulos y paquetes permiten dividir el sistema en partes más pequeñas y organizadas.


# 1.11.1 Modules

## ¿Qué es un módulo?

Un módulo es simplemente un archivo `.py`.

Todo archivo Python es un módulo.


````python
# calculos.py

def calcular_iva(importe):
    return importe * 0.16

def calcular_total(importe):
    return importe + calcular_iva(importe)
````

## Forma tradicional

Sin módulos, todas las funciones viven en el mismo archivo.


## Importando el módulo completo

````python
# main.py

import modulos.calculos as calculos

print(calculos.calcular_iva(1000))
print(calculos.calcular_total(1000))
````

## Ventajas

- Organización.
- Reutilización.
- Separación de responsabilidades.


## Importando funciones específicas

Si solo necesitamos algunas funciones podemos importarlas directamente.


````python
from calculos import calcular_total

print(calcular_total(1000))
````

## Caso de negocio

Imaginemos un sistema comercial dividido en:

- ventas.py
- clientes.py
- reportes.py
- utilidades.py

Cada archivo tiene una responsabilidad específica.


# La variable __name__

Cada módulo contiene una variable especial llamada `__name__`.


````python
print(__name__)
````

## ¿Qué problema resuelve?

Permite distinguir si un archivo:

- Fue ejecutado directamente.
- Fue importado desde otro módulo.


````python
def prueba():
    print("Ejecutando pruebas")

if __name__ == "__main__":
    prueba()
````

## Análisis

Esta estructura es muy común para:

- Pruebas rápidas.
- Ejemplos.
- Código de validación.


# 1.11.2 Packages

## ¿Qué es un paquete?

Un paquete es una carpeta que contiene módulos relacionados.

Tradicionalmente incluye un archivo llamado:

```text
__init__.py
```


## Caso de negocio

Supongamos un sistema de gestión comercial.

```text
proyecto/

ventas/
    __init__.py
    pedidos.py
    facturacion.py

clientes/
    __init__.py
    contactos.py
    creditos.py

reportes/
    __init__.py
    dashboard.py
```


## ¿Qué hace __init__.py?

Puede:

- Estar vacío.
- Inicializar el paquete.
- Definir elementos públicos.


````python
# ventas/__init__.py

VERSION = "1.0" 
````

## Ventajas de los paquetes

- Agrupan módulos relacionados.
- Facilitan el mantenimiento.
- Escalan mejor en proyectos grandes.


# 1.11.3 Search Path for Modules and Packages

## ¿Cómo encuentra Python un módulo?

Cuando ejecutamos:

```python
import modulo
```

Python busca en varios lugares.


### Orden de búsqueda

1. Directorio actual.
2. Biblioteca estándar.
3. Rutas definidas en `sys.path`.


````python
import sys

for ruta in sys.path:
    print(ruta)
````

## Caso de negocio

Supongamos que un equipo comparte librerías corporativas.

Estas pueden agregarse al path para que varios proyectos las reutilicen.


````python
import sys

print(sys.path)
````

## Buenas prácticas

- Evita archivos gigantes.
- Agrupa funcionalidades relacionadas.
- Usa nombres descriptivos.
- Mantén una estructura consistente.
- Separa lógica de negocio, datos y utilerías.
